```python
#import regular libraries
import pandas as pd
import numpy as np
import sklearn

#Viz Imports
import matplotlib.pyplot as plt
import seaborn as sns

#preprocessing imports
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split

#for Training
from sklearn.metrics import mean_squared_error
from sklearn.ensemble import RandomForestRegressor

```

```python
#Load and read the datasets
#indexcol=0 - ID column to index the dataframe
X_train=pd.read_csv('/kaggle/input/30-days-of-ml/train.csv',index_col=0)
y_test=pd.read_csv('/kaggle/input/30-days-of-ml/test.csv',index_col=0)
```

```python
X_train.head()
```

```python
print('There are '+ str(X_train.shape[0])+' rows and '+ str(X_train.shape[1]) +' Columns')
```

```python
#lets look at the columns
X_train.columns
```

Seperate target and feature variables

```python
y=X_train['target']
features=X_train.drop(['target'],axis=1)
```

Prepare the Data

```python
#list of categorical columns
object_cols=[col for col in features.columns if 'cat' in col]

#ordinal Encode categorical columns
X=features.copy()
x_test=y_test.copy()

#initiate ordinal encoder
ordinal_encoder=OrdinalEncoder()
X[object_cols]=ordinal_encoder.fit_transform(features[object_cols])
x_test[object_cols]=ordinal_encoder.transform(y_test[object_cols])

X.head()
```

Break the validation set from the training data

```python
X_train,X_valid,y_train,y_valid=train_test_split(X,y,random_state=0)
```

Train the Model

```python
#Define the model
model=RandomForestRegressor(random_state=1)

#train the model
model.fit(X_train, y_train)
pred_valid=model.predict(X_valid)
print(mean_squared_error(y_valid,pred_valid, squared=False))
```

```python
# Use the model to generate predictions
predictions = model.predict(x_test)

# Save the predictions to a CSV file
output = pd.DataFrame({'Id': x_test.index,
                       'target': predictions})
output.to_csv('submission.csv', index=False)
```

**Which airline should we book, and at what time of day?**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
flights=pd.read_csv("/kaggle/input/datasets/lampubhutia/nyc-flight-delay/flight_data.csv")
flights.head()
```

Groupby Does 3 things in one go:
- Split : Put the rows into piles, one pile per carrier
- apply : compute something on each pile(sum, mean,..)
- combine : glue the answers into one series, indexed by carrier.

```python
flights.groupby("carrier")["dep_delay"].sum().sort_values(ascending=False).head()
```

From the above we can see that,

- Ev has by far the most delay minutes. Is expressJet most likely make you late?

```python
flights.groupby("carrier")["dep_delay"].mean().sort_values(ascending=False).head()
```

A total grows with the number of flights: EV flies 54000 times a year, F9 only 685. To compare groups of different sizes, use a rate or an  average, not sum.
